# Sorting Algorithms

A review of common sorting algorithms, their complexity, mental models, and Python implementations.

The goal is to remember each algorithm from its **core idea**, rather than memorizing the implementation line-by-line.

## Quick Reference

| Algorithm | Average | Worst | Space | Key Idea |
| --- | ---: | ---: | ---: | --- |
| Bubble | O(n²) | O(n²) | O(1) | Swap adjacent inversions |
| Insertion | O(n²) | O(n²) | O(1) | Insert into sorted prefix |
| Selection | O(n²) | O(n²) | O(1) | Find minimum and place it |
| Merge | O(n log n) | O(n log n) | O(n) | Divide, recursively sort, merge |
| Quick | O(n log n) | O(n²) | Implementation-dependent | Partition around a pivot |
| Bucket | Distribution-dependent | O(n²) | O(n + k) | Distribute values into buckets |
| Radix | O(d(n + k)) | O(d(n + k)) | O(n + k) | Stable sort by each digit |

Where:

- `n` = number of elements
- `k` = number of buckets / possible digit values
- `d` = number of digits in the largest number

# 1. Bubble Sort

## Mental Model

> **Compare neighbors → swap if out of order → largest value bubbles to the end.**

Example:

```text
[5, 2, 4, 1]

5 > 2 → swap
[2, 5, 4, 1]

5 > 4 → swap
[2, 4, 5, 1]

5 > 1 → swap
[2, 4, 1, 5]
            ↑
        finalized
```

After every pass, one more element at the end is in its final position.

## How to Reconstruct It

1. Make multiple passes through the array.
2. Compare `nums[j]` with `nums[j + 1]`.
3. Swap if they are out of order.
4. Each pass can ignore one additional element at the end.

## Implementation

```python
def bubble_sort(nums):
    n = len(nums)

    for i in range(n):
        for j in range(n - 1 - i):
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]

    return nums
```

## Complexity

- Average: `O(n²)`
- Worst: `O(n²)`
- Space: `O(1)`

An optimized version can detect when no swaps occur and terminate early, giving `O(n)` best-case time.

# 2. Insertion Sort

## Mental Model

> **Sorting playing cards in your hand.**

Maintain a sorted section on the left and insert each new value into its correct position.

```text
[2, 5, 7 | 3, 6]
             ↑
          current

Move 7 right
Move 5 right

[2, 3, 5, 7 | 6]
```

## How to Reconstruct It

1. Assume the first element is already sorted.
2. Take the next element as `current`.
3. Walk backward through the sorted section.
4. Shift larger values one position to the right.
5. Insert `current` into the opening.

## Implementation

```python
def insertion_sort(nums):
    for i in range(1, len(nums)):
        current = nums[i]
        j = i - 1

        while j >= 0 and nums[j] > current:
            nums[j + 1] = nums[j]
            j -= 1

        nums[j + 1] = current

    return nums
```

## Complexity

- Average: `O(n²)`
- Worst: `O(n²)`
- Best: `O(n)`
- Space: `O(1)`

Insertion Sort performs well on small or nearly sorted datasets.

# 3. Selection Sort

## Mental Model

> **Find the smallest remaining element and put it next.**

```text
[5, 3, 8, 1]
 ↑

Smallest = 1

[1, 3, 8, 5]
    ↑

Smallest remaining = 3
```

The left side gradually becomes finalized.

## How to Reconstruct It

For every position `i`:

1. Assume `i` contains the minimum.
2. Search everything after `i`.
3. Remember the index of the smallest value.
4. Swap it into position `i`.

## Implementation

```python
def selection_sort(nums):
    for i in range(len(nums)):
        min_index = i

        for j in range(i + 1, len(nums)):
            if nums[j] < nums[min_index]:
                min_index = j

        nums[i], nums[min_index] = nums[min_index], nums[i]

    return nums
```

## Complexity

- Average: `O(n²)`
- Worst: `O(n²)`
- Space: `O(1)`

# 4. Merge Sort

## Mental Model

> **Split → recursively sort → merge.**

```text
            [5, 2, 8, 1]
             /        \
         [5, 2]      [8, 1]
         /   \        /   \
       [5]   [2]    [8]   [1]

         ↓ merge      ↓ merge

         [2, 5]      [1, 8]

               ↓ merge

            [1, 2, 5, 8]
```

## How to Reconstruct It

### Part 1 — Recursion

1. If the array has at most one element, it is already sorted.
2. Find the middle.
3. Ask Merge Sort to give you the **left half sorted**.
4. Ask Merge Sort to give you the **right half sorted**.

```text
BASE CASE
↓
SPLIT
↓
RECURSE LEFT
↓
RECURSE RIGHT
```

### Part 2 — Merge

Now you have **two sorted arrays**.

Use two pointers:

```text
left  = [2, 5, 9]
         ↑

right = [1, 3, 8]
         ↑
```

Compare the front values and append the smaller one.

Repeat until one side is exhausted, then append the remainder.

```text
COMPARE
↓
TAKE SMALLER
↓
MOVE POINTER
↓
APPEND REMAINDER
```

## Implementation

```python
def merge_sort(nums):
    # Base case
    if len(nums) <= 1:
        return nums

    # Split
    mid = len(nums) // 2

    # Recursively get both halves sorted
    left = merge_sort(nums[:mid])
    right = merge_sort(nums[mid:])

    # Merge the sorted halves
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    # One of these will be empty
    result.extend(left[i:])
    result.extend(right[j:])

    return result
```

## Why O(n log n)?

Splitting produces approximately `log n` levels:

```text
              n
          /       \
        n/2       n/2
       /  \       /  \
     n/4  n/4   n/4  n/4
            ...
```

Across each level, merging processes `n` elements.

Therefore:

```text
O(n) work per level
×
O(log n) levels
=
O(n log n)
```

## Complexity

- Average: `O(n log n)`
- Worst: `O(n log n)`
- Space: `O(n)`

### Memory Sentence

> **Give me the left sorted, give me the right sorted, then I'll merge them.**

# 5. Quick Sort

## Mental Model

> **Choose pivot → partition → recursively sort both sides.**

Choose a pivot:

```text
[5, 2, 8, 1, 4]
             ↑
           pivot
```

Partition:

```text
smaller/equal     pivot     larger
[2, 1]              4       [5, 8]
```

Then recursively Quick Sort both sides.

## How to Reconstruct It

1. Base case: zero or one element is already sorted.
2. Choose a pivot.
3. Create a section for values `<= pivot`.
4. Create a section for values `> pivot`.
5. Recursively sort both sections.
6. Combine:

```text
sorted left + pivot + sorted right
```

## Simple Implementation

```python
def quick_sort(nums):
    if len(nums) <= 1:
        return nums

    pivot = nums[-1]

    left = []
    right = []

    for num in nums[:-1]:
        if num <= pivot:
            left.append(num)
        else:
            right.append(num)

    return quick_sort(left) + [pivot] + quick_sort(right)
```

This implementation prioritizes understanding over being in-place.

## Why Can Quick Sort Become O(n²)?

If the pivot repeatedly creates balanced partitions:

```text
n
↓
n/2 + n/2
↓
n/4 + n/4 + ...
```

we get approximately `O(log n)` recursion levels and `O(n)` partitioning per level:

```text
O(n log n)
```

But a bad pivot can repeatedly produce:

```text
n
↓
n - 1
↓
n - 2
↓
n - 3
↓
...
```

For example, choosing the final element as the pivot on an already sorted array.

That produces worst-case:

```text
O(n²)
```

## Complexity

- Average: `O(n log n)`
- Worst: `O(n²)`
- Space: implementation-dependent

### Memory Sentence

> **Pivot → smaller left → larger right → recurse.**

# 6. Bucket Sort

## Mental Model

> **Distribute → process each bucket → combine.**

Instead of directly comparing every value, distribute values into groups based on their range.

```text
Numbers
   ↓
[ Bucket 0 ]
[ Bucket 1 ]
[ Bucket 2 ]
[ Bucket 3 ]
   ↓
Sort each bucket
   ↓
Combine
```

Example:

```text
[42, 32, 33, 52, 37, 47, 51]

↓ distribute

Bucket 0: smaller values
Bucket 1: ...
Bucket 2: middle values
Bucket 3: ...
Bucket 4: larger values

↓ sort buckets

↓ concatenate

[32, 33, 37, 42, 47, 51, 52]
```

## How to Reconstruct It

1. Determine the range of the data.
2. Create buckets.
3. Map each value into an appropriate bucket.
4. Sort/process each bucket.
5. Concatenate the buckets in order.

## Implementation

```python
def bucket_sort(nums):
    if len(nums) <= 1:
        return nums

    min_val = min(nums)
    max_val = max(nums)

    value_range = max_val - min_val

    if value_range == 0:
        return nums

    bucket_count = len(nums)
    buckets = [[] for _ in range(bucket_count)]

    # Distribute
    for num in nums:
        index = (
            (num - min_val)
            * (bucket_count - 1)
            // value_range
        )
        buckets[index].append(num)

    # Sort each bucket
    for bucket in buckets:
        bucket.sort()

    # Combine
    result = []

    for bucket in buckets:
        result.extend(bucket)

    return result
```

## Important LeetCode Pattern

Sometimes the problem gives you a natural bucket index.

For **Top K Frequent Elements**, frequency can become the bucket index:

```text
value → frequency

1 → 3
2 → 2
4 → 1
```

Create:

```text
frequency:

0     1     2     3
↓     ↓     ↓     ↓
[]   [4]   [2]   [1]
```

This avoids sorting all `(value, frequency)` pairs.

## Complexity

Bucket Sort's performance depends heavily on:

- Number of buckets
- Distribution of the input
- Algorithm used to sort individual buckets

Under favorable assumptions, it can approach:

```text
O(n + k)
```

If values cluster badly into one bucket and that bucket requires expensive sorting, performance can degrade significantly, potentially to `O(n²)` depending on the implementation.

### Memory Sentence

> **Put similar values together → sort the small groups → combine.**

# 7. Radix Sort

## Mental Model

> **Sort by one digit at a time, from least significant to most significant.**

For:

```text
170
045
075
090
002
024
802
066
```

First sort by the **ones digit**:

```text
170 → 0
090 → 0
002 → 2
802 → 2
024 → 4
045 → 5
075 → 5
066 → 6
```

Then sort by the **tens digit** while preserving the previous ordering.

Then hundreds.

```text
ONES
  ↓
TENS
  ↓
HUNDREDS
  ↓
SORTED
```

## Why Stability Matters

Each digit pass must be **stable**.

If two numbers have the same current digit, they must preserve their relative ordering from the previous pass.

Otherwise, the work done by earlier digit passes can be destroyed.

## How to Extract a Digit

Use:

```python
digit = (num // exp) % 10
```

where:

```text
exp = 1     → ones
exp = 10    → tens
exp = 100   → hundreds
```

Example with `347`:

```text
(347 // 1) % 10   = 7
(347 // 10) % 10  = 4
(347 // 100) % 10 = 3
```

## Implementation

This version handles non-negative integers:

```python
def radix_sort(nums):
    if len(nums) <= 1:
        return nums

    maximum = max(nums)
    exp = 1

    while maximum // exp > 0:
        buckets = [[] for _ in range(10)]

        # Stable distribution into digit buckets
        for num in nums:
            digit = (num // exp) % 10
            buckets[digit].append(num)

        # Rebuild in bucket order
        nums = []

        for bucket in buckets:
            nums.extend(bucket)

        # Move to the next digit
        exp *= 10

    return nums
```

## Complexity

For `d` digits and `k` possible digit values:

- Time: `O(d(n + k))`
- Space: `O(n + k)`

For decimal numbers:

```text
k = 10
```

Radix Sort is a **non-comparison sorting algorithm**.

### Memory Sentence

> **Ones → tens → hundreds, preserving previous ordering.**

# Comparison vs Non-Comparison Sorting

### Comparison Sorts

These determine ordering by comparing values:

```text
Bubble
Insertion
Selection
Merge
Quick
```

General comparison-based sorting cannot beat `O(n log n)` asymptotically in the general case.

### Non-Comparison Sorts

These exploit structure in the values:

```text
Bucket
Radix
```

They can sometimes outperform `O(n log n)` because they are not restricted to pairwise comparisons.

# Interview Recall

Before writing code, explain the algorithm in one sentence:

| Algorithm | One-Sentence Explanation |
| --- | --- |
| Bubble | Repeatedly swap adjacent inversions so large values bubble toward the end. |
| Insertion | Maintain a sorted prefix and insert each new element into its correct position. |
| Selection | Repeatedly find the smallest remaining value and place it next. |
| Merge | Recursively sort two halves, then merge the sorted halves using two pointers. |
| Quick | Partition values around a pivot and recursively sort the partitions. |
| Bucket | Distribute values into ordered groups, process each group, then concatenate them. |
| Radix | Stable-sort values one digit at a time from least significant to most significant. |

The implementation should follow naturally from the explanation.

| Algorithm |                 Average |       Worst |                    Space | Key idea                 |
| --------- | ----------------------: | ----------: | -----------------------: | -----------------------  |
| Bubble    |                   O(n²) |       O(n²) |                     O(1) | Swap adjacent elements   |
| Merge     |              O(n log n) |  O(n log n) |                     O(n) | Divide + merge           |
| Quick     |              O(n log n) |       O(n²) | implementation-dependent | Partition around pivot   |
| Radix     |               O(d(n+k)) |   O(d(n+k)) |                   O(n+k) | Sort digits/positions    |
| Bucket    | depends on distribution | O(n²) worst |                   O(n+k) | Distribute into buckets  |
| Selection |                   O(n²) |       O(n²) |                     O(1) | Find smallest place & it |
| Insertion |                   O(n²) |       O(n²) |                     O(1) | Sorting playing cards    |


````python
def bubble_sort(nums):
    n = len(nums)

    for i in range(n):
        for j in range(n - 1 - i):
            if nums[j] > nums[j + 1]:
                nums[j], nums[j + 1] = nums[j + 1], nums[j]

    return nums

def insertion_sort(nums):
    for i in range(1, len(nums)):
        current = nums[i]
        j = i - 1

        while j >= 0 and nums[j] > current:
            nums[j + 1] = nums[j]
            j -= 1

        nums[j + 1] = current

    return nums

def selection_sort(nums):
    for i in range(len(nums)):
        min_index = i

        for j in range(i + 1, len(nums)):
            if nums[j] < nums[min_index]:
                min_index = j

        nums[i], nums[min_index] = nums[min_index], nums[i] 
    
    return nums

def merge_sort(nums):
    if len(nums) <= 1:
        return nums

    mid = len(nums) // 2

    left_side = nums[:mid]
    right_side = nums[mid:]

    left = merge_sort(left_side)
    right = merge_sort(right_side)

    result = []

    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] < right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1
    
    result.extend(left[i:])
    result.extend(right[j:])

    return result

def quick_sort(nums):
    if len(nums) <= 1:
        return nums
    
    pivot = nums[-1]

    left = []
    right = []

    for num in nums[-1]:
        if num <= pivot:
            left.append(num)
        else:
            right.append(num)
    
    return quick_sort(left) + [pivot] + quick_sort(right)


def bucket_sort(nums):
    if len(nums) <= 1:
        return nums

    # 1. Find the range of values
    min_val = min(nums)
    max_val = max(nums)

    # 2. Create buckets
    bucket_count = len(nums)
    buckets = [[] for _ in range(bucket_count)]

    # 3. Distribute each number into a bucket
    value_range = max_val - min_val

    if value_range == 0:
        return nums

    for num in nums:
        index = (num - min_val) * (bucket_count - 1) // value_range
        buckets[index].append(num)

    # 4. Sort each individual bucket
    for bucket in buckets:
        bucket.sort()

    # 5. Combine the buckets
    result = []

    for bucket in buckets:
        result.extend(bucket)

    return result

def radix_sort(nums):
    radix_array = [[], [], [], [], [], [], [], [], [], [], []] # digits 0, 1, 2... 10

    maximum = max(nums)
    exp = 1

    while maximum // exp > 0:
        while len(nums) > 0:
            val = nums.pop()
            radix_index = (val // exp) % 10
            radix_array[radix_index].append(val)
        
        for bucket in radix_array:
            while len(bucket) > 0:
                val = bucket.pop()
                nums.append(val)
        
        exp *= 10
    
    return nums
````

````python
import random

# Runner
algorithms = {
    "bubble": bubble_sort,
    "insertion": insertion_sort,
    "selection": selection_sort,
    "merge": merge_sort,
    "quick": quick_sort,
    "bucket": bucket_sort,
    "radix": radix_sort
}

print("Available algorithms:")
for name in algorithms:
    print("-", name)

choice = input("\nChoose a sorting algorithm: ").lower()

if choice not in algorithms:
    print("Invalid algorithm.")
else:
    sort_function = algorithms[choice]

    for i in range(10):
        nums = [random.randint(0, 100) for _ in range(10)]

        result = sort_function(nums.copy())
        expected = sorted(nums)

        print(f"\nTest {i + 1}")
        print("Before:  ", nums)
        print("After:   ", result)
        print("Expected:", expected)
        print("PASS" if result == expected else "FAIL")
````
