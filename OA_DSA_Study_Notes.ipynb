# 3. Hash Maps / Sets

## Two Sum — Easy

**Pattern:** Hash map

**Recognition clues:** Need two values that sum to a target, and you want better than O(n²).

**Core idea:** While scanning, store each number's index. For `x`, ask whether `target - x` was already seen.

**Complexity:** Time O(n), space O(n).

**Common trap:** Do the lookup before inserting the current element so you do not reuse the same index.

````python
class Solution:
    def twoSum(self, nums, target):
        seen = {}

        for i, x in enumerate(nums):
            need = target - x
            if need in seen:
                return [seen[need], i]
            seen[x] = i
````

## Contains Duplicate — Easy

**Pattern:** Set

**Recognition clues:** Question asks whether any value appears more than once.

**Core idea:** A set tracks values already encountered.

**Complexity:** Time O(n), space O(n).

**Common trap:** Do not overcomplicate this with sorting unless space constraints specifically matter.

````python
class Solution:
    def containsDuplicate(self, nums):
        seen = set()

        for x in nums:
            if x in seen:
                return True
            seen.add(x)

        return False

    # Other Solution:
    def containsDuplicateOther(self, nums):
        return len(nums) != len(set(nums))
````

## Valid Anagram — Easy

**Pattern:** Frequency counting

**Recognition clues:** Need to check whether two strings contain exactly the same character counts.

**Core idea:** Compare frequency maps (or Counters).

**Complexity:** Time O(n), space O(k) for the alphabet / distinct characters.

**Common trap:** Equal length alone is not enough.

````python
class Solution:
    def isAnagram(self, s, t):
        return Counter(s) == Counter(t)

# Other solution, build the frequency maps manually
class Solution:
    def isAnagramOther(self, s, t):
        if len(s) != len(t):
            return False

        freq_s = {}
        freq_t = {}

        for char in s:
            freq_s[char] = freq_s.get(char, 0) + 1

        for char in t:
            freq_t[char] = freq_t.get(char, 0) + 1

        return freq_s == freq_t

# Note: Counter has the same implementation as freq counting
s = "aabbc"

freq = {}

for char in s:
    freq[char] = freq.get(char, 0) + 1

Counter("s") == freq
````

## Group Anagrams — Medium

**Pattern:** Hash map + canonical key

**Recognition clues:** Need to group strings by an equivalence relation.

**Core idea:** Convert each word into a canonical representation. A 26-count tuple is a good key when input is lowercase English letters.

**Complexity:** Time O(total characters), space O(total characters).

**Common trap:** A list cannot be a dictionary key; convert the count array to a tuple.

````python
class Solution:
    def groupAnagrams(self, strs):
        groups = defaultdict(list)

        for word in strs:
            count = [0] * 26

            for ch in word:
                count[ord(ch) - ord('a')] += 1

            groups[tuple(count)].append(word)

        return list(groups.values())

    def groupAnagramsBetterSpecificSolution(self, strs):
        groups = defaultdict(list)

        for word in strs:
            key = ''.join(sorted(word))
            groups[key].append(word)

        return list(groups.values())
````

## Top K Frequent Elements — Medium

**Pattern:** Frequency map + heap

**Recognition clues:** Need the k most frequent values.

**Core idea:** Count frequencies, then keep a heap of size at most k.

**Complexity:** Time O(n log k), space O(n).

**Common trap:** Heap entries are `(frequency, value)` so the smallest frequency is removed first.

````python
class Solution:
    def topKFrequent(self, nums, k):
        counts = Counter(nums)
        heap = []

        for value, freq in counts.items():
            heapq.heappush(heap, (freq, value))
            if len(heap) > k:
                heapq.heappop(heap)

        return [value for freq, value in heap]
````

# 4. Two Pointers

## Valid Palindrome — Easy

**Pattern:** Two pointers

**Recognition clues:** Compare characters symmetrically while ignoring punctuation / case.

**Core idea:** Move inward; skip non-alphanumeric characters.

**Complexity:** Time O(n), space O(1) excluding normalized copies.

**Common trap:** Increment/decrement carefully after skipping characters.

````python
class Solution:
    def isPalindrome(self, s):
        l, r = 0, len(s) - 1

        while l < r:
            while l < r and not s[l].isalnum():
                l += 1
            while l < r and not s[r].isalnum():
                r -= 1

            if s[l].lower() != s[r].lower():
                return False

            l += 1
            r -= 1

        return True
````

## Two Sum II — Medium

**Pattern:** Two pointers on sorted array

**Recognition clues:** Array is sorted and you need a pair meeting a target.

**Core idea:** If sum is too small, move left pointer right; if too large, move right pointer left.

**Complexity:** Time O(n), space O(1).

**Common trap:** LeetCode's answer is 1-indexed for this problem.

````python
class Solution:
    def twoSum(self, numbers, target):
        l, r = 0, len(numbers) - 1

        while l < r:
            total = numbers[l] + numbers[r]

            if total == target:
                return [l + 1, r + 1]
            elif total < target:
                l += 1
            else:
                r -= 1
````

## Container With Most Water — Medium

**Pattern:** Two pointers / greedy

**Recognition clues:** Area uses two endpoints; width shrinks whenever a pointer moves.

**Core idea:** Always move the shorter wall because the shorter wall limits the height.

**Complexity:** Time O(n), space O(1).

**Common trap:** Moving the taller wall cannot improve the current height limitation.

````python
class Solution:
    def maxArea(self, height):
        l, r = 0, len(height) - 1
        best = 0

        while l < r:
            area = (r - l) * min(height[l], height[r])
            best = max(best, area)

            if height[l] < height[r]:
                l += 1
            else:
                r -= 1

        return best
````

## 3Sum — Medium

**Pattern:** Sorting + two pointers

**Recognition clues:** Need unique triplets whose sum is zero.

**Core idea:** Sort. Fix one number, then solve a two-sum problem on the suffix.

**Complexity:** Time O(n²), extra space O(1) aside from output / sorting implementation.

**Common trap:** Skip duplicates for the fixed element and after finding a valid pair.

````python
class Solution:
    def threeSum(self, nums):
        nums.sort()
        result = []

        for i in range(len(nums)):
            if i > 0 and nums[i] == nums[i - 1]:
                continue
            if nums[i] > 0:
                break

            l, r = i + 1, len(nums) - 1

            while l < r:
                total = nums[i] + nums[l] + nums[r]

                if total < 0:
                    l += 1
                elif total > 0:
                    r -= 1
                else:
                    result.append([nums[i], nums[l], nums[r]])
                    l += 1
                    r -= 1

                    while l < r and nums[l] == nums[l - 1]:
                        l += 1

        return result
````

# 5. Sliding Window

## Best Time to Buy and Sell Stock — Easy

**Pattern:** Running minimum / window

**Recognition clues:** Need best future sell price relative to an earlier buy.

**Core idea:** Track the cheapest price seen so far and calculate profit if selling today.

**Complexity:** Time O(n), space O(1).

**Common trap:** Buy must occur before sell; do not simply take global min and max independently.

````python
class Solution:
    def maxProfit(self, prices):
        min_price = float('inf')
        best = 0

        for price in prices:
            min_price = min(min_price, price)
            best = max(best, price - min_price)

        return best
````

## Longest Substring Without Repeating Characters — Medium

**Pattern:** Sliding window + set

**Recognition clues:** Longest contiguous substring with a uniqueness constraint.

**Core idea:** Expand right. While the new character duplicates something in the window, remove from the left.

**Complexity:** Time O(n), space O(k).

**Common trap:** Use `while`, not `if`, because more than one left shift may be needed.

````python
class Solution:
    def lengthOfLongestSubstring(self, s):
        seen = set()
        l = 0
        best = 0

        for r, ch in enumerate(s):
            while ch in seen:
                seen.remove(s[l])
                l += 1

            seen.add(ch)
            best = max(best, r - l + 1)

        return best
````

## Longest Repeating Character Replacement — Medium

**Pattern:** Sliding window + frequency

**Recognition clues:** Longest window that can become all one character after at most k replacements.

**Core idea:** For a window of length L, replacements needed are `L - max_frequency_in_window`.

**Complexity:** Time O(n), space O(alphabet).

**Common trap:** The stored `max_freq` can be stale; that is okay for the standard O(n) solution because it never causes an invalid best length to be missed.

````python
class Solution:
    def characterReplacement(self, s, k):
        counts = defaultdict(int)
        l = 0
        max_freq = 0
        best = 0

        for r, ch in enumerate(s):
            counts[ch] += 1
            max_freq = max(max_freq, counts[ch])

            while (r - l + 1) - max_freq > k:
                counts[s[l]] -= 1
                l += 1

            best = max(best, r - l + 1)

        return best
````

## Minimum Size Subarray Sum — Medium

**Pattern:** Sliding window

**Recognition clues:** Positive numbers + smallest contiguous subarray meeting a threshold.

**Core idea:** Expand until sum is large enough, then shrink greedily from the left while still valid.

**Complexity:** Time O(n), space O(1).

**Common trap:** This exact sliding-window logic depends on values being positive.

````python
class Solution:
    def minSubArrayLen(self, target, nums):
        l = 0
        total = 0
        best = float('inf')

        for r, x in enumerate(nums):
            total += x

            while total >= target:
                best = min(best, r - l + 1)
                total -= nums[l]
                l += 1

        return 0 if best == float('inf') else best
````

## Maximum Number of Robots Within Budget — Hard

**Pattern:** Sliding window + monotonic deque

**Recognition clues:** Window cost combines a sum and the maximum value inside the window.

**Core idea:** Maintain running costs with a sum and maximum charge time with a decreasing deque of indices. Shrink while over budget.

**Complexity:** Time O(n), space O(n).

**Common trap:** Remove deque indices that leave the window; never search linearly for the maximum.

````python
class Solution:
    def maximumRobots(self, chargeTimes, runningCosts, budget):
        q = deque()
        running_sum = 0
        l = 0
        best = 0

        for r in range(len(chargeTimes)):
            running_sum += runningCosts[r]

            while q and chargeTimes[q[-1]] <= chargeTimes[r]:
                q.pop()
            q.append(r)

            while q and chargeTimes[q[0]] + (r - l + 1) * running_sum > budget:
                if q[0] == l:
                    q.popleft()
                running_sum -= runningCosts[l]
                l += 1

            best = max(best, r - l + 1)

        return best
````

# 6. Stacks / Monotonic Stacks

## Valid Parentheses — Easy

**Pattern:** Stack

**Recognition clues:** Nested opening/closing symbols must be matched in reverse order.

**Core idea:** Push opening brackets; each closing bracket must match the stack top.

**Complexity:** Time O(n), space O(n).

**Common trap:** Check whether the stack is empty before reading its top.

````python
class Solution:
    def isValid(self, s):
        pairs = {')': '(', ']': '[', '}': '{'}
        stack = []

        for ch in s:
            if ch in pairs:
                if not stack or stack[-1] != pairs[ch]:
                    return False
                stack.pop()
            else:
                stack.append(ch)

        return not stack
````

## Daily Temperatures — Medium

**Pattern:** Monotonic decreasing stack

**Recognition clues:** For each item, find the next later item that is greater.

**Core idea:** Store unresolved indices in a decreasing stack. A warmer temperature resolves earlier indices.

**Complexity:** Time O(n), space O(n).

**Common trap:** Store indices, not only temperatures, because the answer is a distance.

````python
class Solution:
    def dailyTemperatures(self, temperatures):
        answer = [0] * len(temperatures)
        stack = []

        for i, temp in enumerate(temperatures):
            while stack and temperatures[stack[-1]] < temp:
                j = stack.pop()
                answer[j] = i - j
            stack.append(i)

        return answer
````

## Min Stack — Medium

**Pattern:** Stack with auxiliary minimum

**Recognition clues:** Stack must support push/pop/top and minimum in O(1).

**Core idea:** Store the minimum associated with every stack depth.

**Complexity:** All operations O(1), space O(n).

**Common trap:** Recomputing min after every pop would make pop O(n).

````python
class MinStack:
    def __init__(self):
        self.stack = []

    def push(self, val):
        current_min = val if not self.stack else min(val, self.stack[-1][1])
        self.stack.append((val, current_min))

    def pop(self):
        self.stack.pop()

    def top(self):
        return self.stack[-1][0]

    def getMin(self):
        return self.stack[-1][1]
````

## Car Fleet — Medium

**Pattern:** Sorting + monotonic reasoning

**Recognition clues:** Cars move toward one target and faster cars can catch slower cars ahead.

**Core idea:** Sort by starting position descending. Compute each car's arrival time; a new fleet forms only when its time exceeds the fleet ahead.

**Complexity:** Time O(n log n), space O(n).

**Common trap:** Process closest-to-target first; order is essential.

````python
class Solution:
    def carFleet(self, target, position, speed):
        cars = sorted(zip(position, speed), reverse=True)
        fleets = 0
        slowest_time = 0

        for pos, spd in cars:
            time = (target - pos) / spd

            if time > slowest_time:
                fleets += 1
                slowest_time = time

        return fleets
````

# 7. Linked Lists

## Reverse Linked List — Easy

**Pattern:** Pointer manipulation

**Recognition clues:** Need to reverse links in-place.

**Core idea:** Carry `prev`, `curr`, and temporarily save `curr.next` before rewiring.

**Complexity:** Time O(n), space O(1).

**Common trap:** If you overwrite `curr.next` before saving it, you lose the rest of the list.

````python
class Solution:
    def reverseList(self, head):
        prev = None
        curr = head

        while curr:
            nxt = curr.next
            curr.next = prev
            prev = curr
            curr = nxt

        return prev
````

## Linked List Cycle — Easy

**Pattern:** Fast & slow pointers (Floyd's)

**Recognition clues:** Need to detect repeated traversal without extra memory.

**Core idea:** Slow moves 1 step; fast moves 2. If there is a cycle they eventually meet.

**Complexity:** Time O(n), space O(1).

**Common trap:** Check `fast` and `fast.next` before advancing two steps.

````python
class Solution:
    def hasCycle(self, head):
        slow = fast = head

        while fast and fast.next:
            slow = slow.next
            fast = fast.next.next

            if slow is fast:
                return True

        return False
````

## Merge Two Sorted Lists — Easy

**Pattern:** Two pointers + dummy node

**Recognition clues:** Merge two sorted linked lists.

**Core idea:** Use a dummy head and repeatedly attach the smaller current node.

**Complexity:** Time O(n + m), space O(1) excluding output links.

**Common trap:** Remember to attach whichever list remains after one becomes empty.

````python
class Solution:
    def mergeTwoLists(self, list1, list2):
        dummy = ListNode()
        tail = dummy

        while list1 and list2:
            if list1.val <= list2.val:
                tail.next = list1
                list1 = list1.next
            else:
                tail.next = list2
                list2 = list2.next
            tail = tail.next

        tail.next = list1 if list1 else list2
        return dummy.next
````

## Remove Nth Node From End of List — Medium

**Pattern:** Two pointers with gap

**Recognition clues:** Need a position measured from the end in one pass.

**Core idea:** Use a dummy node. Move fast n+1 steps ahead of slow, then advance together.

**Complexity:** Time O(n), space O(1).

**Common trap:** The dummy node cleanly handles removing the head.

````python
class Solution:
    def removeNthFromEnd(self, head, n):
        dummy = ListNode(0, head)
        slow = fast = dummy

        for _ in range(n + 1):
            fast = fast.next

        while fast:
            slow = slow.next
            fast = fast.next

        slow.next = slow.next.next
        return dummy.next
````

# 8. Binary Search

## Binary Search — Easy

**Pattern:** Binary search

**Recognition clues:** Sorted array and exact target lookup.

**Core idea:** Compare midpoint and discard half the search space.

**Complexity:** Time O(log n), space O(1).

**Common trap:** Choose one boundary convention and stay consistent.

````python
class Solution:
    def search(self, nums, target):
        l, r = 0, len(nums) - 1

        while l <= r:
            m = l + (r - l) // 2

            if nums[m] == target:
                return m
            if nums[m] < target:
                l = m + 1
            else:
                r = m - 1

        return -1
````

## Koko Eating Bananas — Medium

**Pattern:** Binary search on answer

**Recognition clues:** Need the minimum feasible rate; feasibility is monotonic.

**Core idea:** Binary-search eating speed from 1 to max pile. For each speed, compute required hours.

**Complexity:** Time O(n log M), space O(1), where M=max(piles).

**Common trap:** The feasibility test uses ceiling division.

````python
class Solution:
    def minEatingSpeed(self, piles, h):
        l, r = 1, max(piles)

        while l < r:
            speed = (l + r) // 2
            hours = sum((pile + speed - 1) // speed for pile in piles)

            if hours <= h:
                r = speed
            else:
                l = speed + 1

        return l
````

## Search in Rotated Sorted Array — Medium

**Pattern:** Modified binary search

**Recognition clues:** Sorted array has been rotated, but one half around every midpoint is still sorted.

**Core idea:** Determine which half is sorted, then decide whether target lies inside that half.

**Complexity:** Time O(log n), space O(1).

**Common trap:** Use inclusive comparisons consistently when testing target range.

````python
class Solution:
    def search(self, nums, target):
        l, r = 0, len(nums) - 1

        while l <= r:
            m = (l + r) // 2

            if nums[m] == target:
                return m

            if nums[l] <= nums[m]:  # left half sorted
                if nums[l] <= target < nums[m]:
                    r = m - 1
                else:
                    l = m + 1
            else:  # right half sorted
                if nums[m] < target <= nums[r]:
                    l = m + 1
                else:
                    r = m - 1

        return -1
````

# 9. Trees

## Maximum Depth of Binary Tree — Easy

**Pattern:** Tree DFS

**Recognition clues:** Need a value based on child subtrees.

**Core idea:** Depth at a node is 1 + max(left depth, right depth).

**Complexity:** Time O(n), recursion space O(h).

**Common trap:** Base case for a missing node is depth 0.

````python
class Solution:
    def maxDepth(self, root):
        if not root:
            return 0

        return 1 + max(self.maxDepth(root.left), self.maxDepth(root.right))
````

## Binary Tree Level Order Traversal — Medium

**Pattern:** BFS

**Recognition clues:** Need nodes grouped by depth / level.

**Core idea:** Process exactly the current queue length for each level.

**Complexity:** Time O(n), space O(w) where w is max tree width.

**Common trap:** If you do not snapshot level size, children may be mixed into the same level.

````python
class Solution:
    def levelOrder(self, root):
        if not root:
            return []

        result = []
        q = deque([root])

        while q:
            level = []

            for _ in range(len(q)):
                node = q.popleft()
                level.append(node.val)

                if node.left:
                    q.append(node.left)
                if node.right:
                    q.append(node.right)

            result.append(level)

        return result
````

## Invert Binary Tree — Easy

**Pattern:** Tree DFS

**Recognition clues:** Need to recursively swap symmetric structure.

**Core idea:** Swap each node's left/right children, then recurse.

**Complexity:** Time O(n), space O(h).

**Common trap:** Return the root after mutation.

````python
class Solution:
    def invertTree(self, root):
        if not root:
            return None

        root.left, root.right = root.right, root.left
        self.invertTree(root.left)
        self.invertTree(root.right)

        return root
````

## Same Tree — Easy

**Pattern:** Parallel tree DFS

**Recognition clues:** Need structural equality and value equality.

**Core idea:** Compare both nodes at the same position recursively.

**Complexity:** Time O(n), space O(h).

**Common trap:** Handle `both None` separately from `only one None`.

````python
class Solution:
    def isSameTree(self, p, q):
        if not p and not q:
            return True
        if not p or not q or p.val != q.val:
            return False

        return self.isSameTree(p.left, q.left) and self.isSameTree(p.right, q.right)
````

## Validate Binary Search Tree — Medium

**Pattern:** DFS with bounds

**Recognition clues:** Every node must satisfy constraints inherited from all ancestors, not only its parent.

**Core idea:** Carry an allowed `(low, high)` range down the tree.

**Complexity:** Time O(n), space O(h).

**Common trap:** Checking only immediate children is insufficient.

````python
class Solution:
    def isValidBST(self, root):
        def valid(node, low, high):
            if not node:
                return True

            if not (low < node.val < high):
                return False

            return valid(node.left, low, node.val) and valid(node.right, node.val, high)

        return valid(root, float('-inf'), float('inf'))
````

## Lowest Common Ancestor of a Binary Search Tree — Medium

**Pattern:** BST property

**Recognition clues:** Need shared ancestor in a BST.

**Core idea:** If both targets are smaller, go left; if both larger, go right; otherwise current node is the split point.

**Complexity:** Time O(h), space O(1) iterative.

**Common trap:** Use BST ordering; a general binary-tree LCA requires a different approach.

````python
class Solution:
    def lowestCommonAncestor(self, root, p, q):
        node = root

        while node:
            if p.val < node.val and q.val < node.val:
                node = node.left
            elif p.val > node.val and q.val > node.val:
                node = node.right
            else:
                return node
````

## Graph reminders

- **DFS/BFS traversal:** O(V + E)
- **Grid traversal:** O(rows × cols)
- Mark a node visited **when you enqueue / enter it**, not after processing, to avoid duplicate work.
- A 2-D grid is an implicit graph: neighbors are usually up/down/left/right.
- BFS gives shortest path length in an **unweighted** graph.

## Number of Islands — Medium

**Pattern:** Grid DFS/BFS

**Recognition clues:** Need number of connected components in a grid.

**Core idea:** Whenever an unvisited land cell is found, increment count and flood-fill the entire island.

**Complexity:** Time O(rows·cols), space O(rows·cols) worst case.

**Common trap:** Mark visited immediately so cells are not revisited.

````python
class Solution:
    def numIslands(self, grid):
        rows, cols = len(grid), len(grid[0])
        islands = 0

        def dfs(r, c):
            if r < 0 or r >= rows or c < 0 or c >= cols or grid[r][c] != '1':
                return

            grid[r][c] = '0'

            dfs(r + 1, c)
            dfs(r - 1, c)
            dfs(r, c + 1)
            dfs(r, c - 1)

        for r in range(rows):
            for c in range(cols):
                if grid[r][c] == '1':
                    islands += 1
                    dfs(r, c)

        return islands
````

## Clone Graph — Medium

**Pattern:** Graph DFS + map

**Recognition clues:** Need a deep copy of a graph that may contain cycles.

**Core idea:** Map each original node to its clone. If already cloned, reuse it.

**Complexity:** Time O(V + E), space O(V).

**Common trap:** Create/store the clone before recursively cloning neighbors, otherwise cycles recurse forever.

````python
class Solution:
    def cloneGraph(self, node):
        if not node:
            return None

        copies = {}

        def dfs(curr):
            if curr in copies:
                return copies[curr]

            clone = Node(curr.val)
            copies[curr] = clone

            for nei in curr.neighbors:
                clone.neighbors.append(dfs(nei))

            return clone

        return dfs(node)
````

## Flood Fill — Easy

**Pattern:** Grid DFS/BFS

**Recognition clues:** Change every cell connected to a starting cell that has the original color.

**Core idea:** Capture the original color, then traverse same-colored neighbors.

**Complexity:** Time O(rows·cols), space O(rows·cols) worst case.

**Common trap:** If new color equals original color, return early to avoid infinite revisits.

````python
class Solution:
    def floodFill(self, image, sr, sc, color):
        original = image[sr][sc]

        if original == color:
            return image

        rows, cols = len(image), len(image[0])

        def dfs(r, c):
            if r < 0 or r >= rows or c < 0 or c >= cols:
                return
            if image[r][c] != original:
                return

            image[r][c] = color

            dfs(r + 1, c)
            dfs(r - 1, c)
            dfs(r, c + 1)
            dfs(r, c - 1)

        dfs(sr, sc)
        return image
````

## Rotting Oranges — Medium

**Pattern:** Multi-source BFS

**Recognition clues:** Several sources spread one step per minute simultaneously.

**Core idea:** Enqueue every rotten orange first, count fresh oranges, then process BFS in layers.

**Complexity:** Time O(rows·cols), space O(rows·cols).

**Common trap:** Starting BFS from only one rotten orange gives incorrect timing.

````python
class Solution:
    def orangesRotting(self, grid):
        rows, cols = len(grid), len(grid[0])
        q = deque()
        fresh = 0

        for r in range(rows):
            for c in range(cols):
                if grid[r][c] == 2:
                    q.append((r, c))
                elif grid[r][c] == 1:
                    fresh += 1

        minutes = 0
        directions = [(1, 0), (-1, 0), (0, 1), (0, -1)]

        while q and fresh:
            for _ in range(len(q)):
                r, c = q.popleft()

                for dr, dc in directions:
                    nr, nc = r + dr, c + dc

                    if 0 <= nr < rows and 0 <= nc < cols and grid[nr][nc] == 1:
                        grid[nr][nc] = 2
                        fresh -= 1
                        q.append((nr, nc))

            minutes += 1

        return minutes if fresh == 0 else -1
````

## Course Schedule — Medium

**Pattern:** Directed graph / DFS cycle detection

**Recognition clues:** Prerequisites form a directed dependency graph. The question is asking whether every course can eventually be completed, which means checking whether the graph contains a cycle.

**Core idea:** Use DFS and track the state of each course:
- `0` = unvisited
- `1` = currently visiting
- `2` = completely processed

If DFS reaches a course that is already in the `visiting` state, we have returned to a node on our current path, meaning a cycle exists.

If any cycle exists, it is impossible to finish all courses.

**Complexity:** Time O(V + E), space O(V + E).

**Common trap:** A node that has already been fully processed is not a cycle. A cycle only exists when we encounter a node that is currently on the DFS path (`state == 1`).

````python
class Solution:
    def canFinish(self, numCourses, prerequisites):
        graph = [[] for _ in range(numCourses)]

        for course, prereq in prerequisites:
            graph[prereq].append(course)

        # 0 = unvisited
        # 1 = visiting
        # 2 = visited
        state = [0] * numCourses

        def dfs(course):
            # We found a node already on our current path
            if state[course] == 1:
                return False

            # Already checked this course and found no cycle
            if state[course] == 2:
                return True

            state[course] = 1

            for next_course in graph[course]:
                if not dfs(next_course):
                    return False

            state[course] = 2
            return True

        for course in range(numCourses):
            if not dfs(course):
                return False

        return True
````

## Pacific Atlantic Water Flow — Medium

**Pattern:** Reverse graph traversal

**Recognition clues:** Cells can flow downhill to oceans; checking every cell forward repeats huge amounts of work.

**Core idea:** Reverse the direction: start from each ocean's border and move to cells with equal-or-greater heights. Intersect reachable sets.

**Complexity:** Time O(rows·cols), space O(rows·cols).

**Common trap:** The reversed traversal uses `next_height >= current_height`, which is easy to flip accidentally.

````python
class Solution:
    def pacificAtlantic(self, heights):
        rows, cols = len(heights), len(heights[0])
        pacific = set()
        atlantic = set()

        def dfs(r, c, visited, prev_height):
            if (
                r < 0 or r >= rows or c < 0 or c >= cols
                or (r, c) in visited
                or heights[r][c] < prev_height
            ):
                return

            visited.add((r, c))
            h = heights[r][c]

            dfs(r + 1, c, visited, h)
            dfs(r - 1, c, visited, h)
            dfs(r, c + 1, visited, h)
            dfs(r, c - 1, visited, h)

        for c in range(cols):
            dfs(0, c, pacific, heights[0][c])
            dfs(rows - 1, c, atlantic, heights[rows - 1][c])

        for r in range(rows):
            dfs(r, 0, pacific, heights[r][0])
            dfs(r, cols - 1, atlantic, heights[r][cols - 1])

        return [[r, c] for r, c in pacific & atlantic]
````

# 11. Dynamic Programming

## DP decision process

Ask these in order:

1. **What is the state?** What minimum information uniquely describes a subproblem?
2. **What is the recurrence?** How does this state depend on smaller states?
3. **What are the base cases?**
4. **In what order can I compute states?**
5. **Can memory be compressed?**

Common clues: number of ways, minimum cost, maximum profit/value, longest subsequence, choices that branch but revisit the same state.

## Climbing Stairs — Easy

**Pattern:** 1-D DP / Fibonacci

**Recognition clues:** Number of ways to reach position n using steps of size 1 or 2.

**Core idea:** Ways to reach i = ways to i-1 + ways to i-2.

**Complexity:** Time O(n), space O(1).

**Common trap:** Initialize base cases carefully; `n=1` should work.

````python
class Solution:
    def climbStairs(self, n):
        if n <= 2:
            return n

        prev2, prev1 = 1, 2

        for _ in range(3, n + 1):
            prev2, prev1 = prev1, prev1 + prev2

        return prev1
````

## House Robber — Medium

**Pattern:** 1-D DP

**Recognition clues:** At each item, either take it and skip the previous one, or skip it.

**Core idea:** `dp[i] = max(dp[i-1], nums[i] + dp[i-2])`.

**Complexity:** Time O(n), space O(1) with compression.

**Common trap:** Do not greedily choose the locally larger house; choices affect future options.

````python
class Solution:
    def rob(self, nums):
        prev2 = 0
        prev1 = 0

        for money in nums:
            current = max(prev1, prev2 + money)
            prev2, prev1 = prev1, current

        return prev1
````

## Unique Paths — Medium

**Pattern:** 2-D DP

**Recognition clues:** Count ways through a grid where each cell can be reached from top or left.

**Core idea:** `dp[r][c] = dp[r-1][c] + dp[r][c-1]`.

**Complexity:** Time O(mn), space O(n) with row compression.

**Common trap:** First row/column each have exactly one path when there are no obstacles.

````python
class Solution:
    def uniquePaths(self, m, n):
        dp = [1] * n

        for _ in range(1, m):
            for c in range(1, n):
                dp[c] += dp[c - 1]

        return dp[-1]
````

## Minimum Path Sum — Medium

**Pattern:** 2-D DP

**Recognition clues:** Need minimum accumulated cost moving only right/down.

**Core idea:** Each cell's best cost is its own value plus min(top, left).

**Complexity:** Time O(mn), space O(n).

**Common trap:** Edges only have one predecessor; initialize them intentionally.

````python
class Solution:
    def minPathSum(self, grid):
        rows, cols = len(grid), len(grid[0])
        dp = [float('inf')] * cols
        dp[0] = 0

        for r in range(rows):
            dp[0] += grid[r][0]

            for c in range(1, cols):
                dp[c] = grid[r][c] + min(dp[c], dp[c - 1])

        return dp[-1]
````

## Coin Change — Medium

**Pattern:** Unbounded knapsack DP

**Recognition clues:** Need minimum number of items to build an exact amount, and each coin may be reused.

**Core idea:** `dp[a] = 1 + min(dp[a - coin])` for usable coins.

**Complexity:** Time O(amount × number_of_coins), space O(amount).

**Common trap:** Use infinity for unreachable states and convert final infinity to -1.

````python
class Solution:
    def coinChange(self, coins, amount):
        dp = [float('inf')] * (amount + 1)
        dp[0] = 0

        for a in range(1, amount + 1):
            for coin in coins:
                if coin <= a:
                    dp[a] = min(dp[a], 1 + dp[a - coin])

        return -1 if dp[amount] == float('inf') else dp[amount]
````

## Longest Common Subsequence — Medium

**Pattern:** 2-D DP

**Recognition clues:** Need longest sequence preserving order in two strings, but characters need not be contiguous.

**Core idea:** If characters match, take diagonal + 1; otherwise take best of dropping one character from either string.

**Complexity:** Time O(mn), space O(mn), compressible to O(n).

**Common trap:** Subsequence is not substring; do not use a sliding window.

````python
class Solution:
    def longestCommonSubsequence(self, text1, text2):
        m, n = len(text1), len(text2)
        dp = [[0] * (n + 1) for _ in range(m + 1)]

        for i in range(m - 1, -1, -1):
            for j in range(n - 1, -1, -1):
                if text1[i] == text2[j]:
                    dp[i][j] = 1 + dp[i + 1][j + 1]
                else:
                    dp[i][j] = max(dp[i + 1][j], dp[i][j + 1])

        return dp[0][0]
````

## Race Car — Hard

**Pattern:** Shortest-path BFS over state space

**Recognition clues:** Actions transform both position and speed, and the shortest instruction sequence is required.

**Core idea:** Treat `(position, speed)` as a graph state. Accelerate doubles speed; reverse changes direction to ±1. BFS finds minimum commands.

**Complexity:** Pseudo-polynomial state search; practical bound keeps explored positions around the target.

**Common trap:** Without bounding positions / visited states, the search can explode.

````python
class Solution:
    def racecar(self, target):
        q = deque([(0, 1, 0)])  # position, speed, steps
        visited = {(0, 1)}
        bound = 2 * target + 2

        while q:
            pos, speed, steps = q.popleft()

            if pos == target:
                return steps

            # Accelerate
            next_state = (pos + speed, speed * 2)
            if -bound <= next_state[0] <= bound and next_state not in visited:
                visited.add(next_state)
                q.append((*next_state, steps + 1))

            # Reverse
            reverse_speed = -1 if speed > 0 else 1
            reverse_state = (pos, reverse_speed)
            if reverse_state not in visited:
                visited.add(reverse_state)
                q.append((pos, reverse_speed, steps + 1))
````

# 12. Heaps / Priority Queues

## Kth Largest Element — Medium

**Pattern:** Min-heap of size k

**Recognition clues:** Need kth largest without fully sorting all values.

**Core idea:** Keep only the k largest values seen; the heap root is the kth largest.

**Complexity:** Time O(n log k), space O(k).

**Common trap:** A Python min-heap is exactly what you want for a size-k collection of largest values.

````python
class Solution:
    def findKthLargest(self, nums, k):
        heap = []

        for x in nums:
            heapq.heappush(heap, x)
            if len(heap) > k:
                heapq.heappop(heap)

        return heap[0]
````

## Last Stone Weight — Easy

**Pattern:** Max-heap

**Recognition clues:** Repeatedly remove the two largest values.

**Core idea:** Use negated values because Python provides a min-heap.

**Complexity:** Time O(n log n), space O(n).

**Common trap:** Negate both when pushing and when interpreting popped values.

````python
class Solution:
    def lastStoneWeight(self, stones):
        heap = [-x for x in stones]
        heapq.heapify(heap)

        while len(heap) > 1:
            first = -heapq.heappop(heap)
            second = -heapq.heappop(heap)

            if first != second:
                heapq.heappush(heap, -(first - second))

        return -heap[0] if heap else 0
````

## Merge K Sorted Lists — Hard

**Pattern:** Heap + linked lists

**Recognition clues:** Need repeatedly find the smallest current head among k sorted lists.

**Core idea:** Push each non-empty head into a heap, pop smallest, append it, then push that node's successor.

**Complexity:** Time O(N log k), space O(k).

**Common trap:** Python cannot compare ListNode objects on ties, so include a unique counter / list index in heap tuples.

````python
class Solution:
    def mergeKLists(self, lists):
        heap = []

        for i, node in enumerate(lists):
            if node:
                heapq.heappush(heap, (node.val, i, node))

        dummy = ListNode()
        tail = dummy

        while heap:
            _, i, node = heapq.heappop(heap)
            tail.next = node
            tail = tail.next

            if node.next:
                heapq.heappush(heap, (node.next.val, i, node.next))

        return dummy.next
````

## Find the K-Sum of an Array — Hard

**Pattern:** Sorting + heap of subset sums

**Recognition clues:** Need kth largest subsequence sum among exponentially many subsequences.

**Core idea:** Start with sum of all positive values (maximum possible). Convert values to absolute costs and generate the smallest reductions using a heap; kth largest = max_sum - kth-smallest reduction.

**Complexity:** Time O(n log n + k log k), space O(n + k).

**Common trap:** The transformation to absolute values is the key insight; brute-force subsets are impossible for large n.

````python
class Solution:
    def kSum(self, nums, k):
        max_sum = sum(x for x in nums if x > 0)
        costs = sorted(abs(x) for x in nums)

        if k == 1:
            return max_sum

        heap = [(costs[0], 0)]
        reduction = 0

        for _ in range(k - 1):
            reduction, i = heapq.heappop(heap)

            if i + 1 < len(costs):
                # Add next cost
                heapq.heappush(heap, (reduction + costs[i + 1], i + 1))

                # Replace current cost with next cost
                heapq.heappush(
                    heap,
                    (reduction - costs[i] + costs[i + 1], i + 1)
                )

        return max_sum - reduction
````

# 13. Intervals

## Interval recognition rule

Most interval problems become much easier after sorting by **start time**.

Then ask:
- Does the current interval overlap the previous merged interval?
- Do I merge, insert, count, or remove?

## Merge Intervals — Medium

**Pattern:** Sort + sweep

**Recognition clues:** Overlapping ranges need to be combined.

**Core idea:** Sort by start. Merge if current start <= previous merged end.

**Complexity:** Time O(n log n), space O(n) for output.

**Common trap:** Remember that touching endpoints usually count as overlap in this problem.

````python
class Solution:
    def merge(self, intervals):
        intervals.sort(key=lambda x: x[0])
        merged = []

        for start, end in intervals:
            if not merged or start > merged[-1][1]:
                merged.append([start, end])
            else:
                merged[-1][1] = max(merged[-1][1], end)

        return merged
````

## Insert Interval — Medium

**Pattern:** Interval sweep

**Recognition clues:** One interval is inserted into an already sorted non-overlapping list.

**Core idea:** Append intervals fully before the new one, merge all overlaps into the new interval, then append remaining intervals.

**Complexity:** Time O(n), space O(n) for output.

**Common trap:** Use three phases; it avoids lots of fragile casework.

````python
class Solution:
    def insert(self, intervals, newInterval):
        result = []
        i = 0
        n = len(intervals)

        while i < n and intervals[i][1] < newInterval[0]:
            result.append(intervals[i])
            i += 1

        while i < n and intervals[i][0] <= newInterval[1]:
            newInterval[0] = min(newInterval[0], intervals[i][0])
            newInterval[1] = max(newInterval[1], intervals[i][1])
            i += 1

        result.append(newInterval)
        result.extend(intervals[i:])

        return result
````

## Non-overlapping Intervals — Medium

**Pattern:** Greedy intervals

**Recognition clues:** Need minimum removals so remaining intervals do not overlap.

**Core idea:** Sort by end time and greedily keep the interval that ends earliest; it leaves maximum room for future intervals.

**Complexity:** Time O(n log n), space O(1) aside from sorting.

**Common trap:** When overlap occurs, keeping the earlier-ending interval is the greedy choice.

````python
class Solution:
    def eraseOverlapIntervals(self, intervals):
        intervals.sort(key=lambda x: x[1])
        removals = 0
        previous_end = float('-inf')

        for start, end in intervals:
            if start >= previous_end:
                previous_end = end
            else:
                removals += 1

        return removals
````

# 14. Backtracking

## Backtracking mental model

At every recursion level:

1. **Choose**
2. **Explore**
3. **Undo**

The path is shared mutable state, so use `path.copy()` when saving an answer.

## Subsets — Medium

**Pattern:** Backtracking

**Recognition clues:** Need every combination of taking/not-taking values.

**Core idea:** At each position, include or exclude the element. Another clean formulation appends the current path at every recursion node.

**Complexity:** Time O(n·2ⁿ) including output copying, recursion space O(n).

**Common trap:** Forgetting `.copy()` makes every saved result reference the same list.

````python
class Solution:
    def subsets(self, nums):
        result = []
        path = []

        def backtrack(start):
            result.append(path.copy())

            for i in range(start, len(nums)):
                path.append(nums[i])
                backtrack(i + 1)
                path.pop()

        backtrack(0)
        return result
````

## Combination Sum — Medium

**Pattern:** Backtracking with reuse

**Recognition clues:** Need combinations adding to a target; a candidate may be reused.

**Core idea:** Track remaining target. Recurse from current index so combinations are generated without permutation duplicates.

**Complexity:** Exponential in target/candidates; recursion depth up to target/min(candidate).

**Common trap:** Recursive call uses `i`, not `i+1`, because reuse is allowed.

````python
class Solution:
    def combinationSum(self, candidates, target):
        result = []
        path = []

        def backtrack(start, remaining):
            if remaining == 0:
                result.append(path.copy())
                return
            if remaining < 0:
                return

            for i in range(start, len(candidates)):
                path.append(candidates[i])
                backtrack(i, remaining - candidates[i])
                path.pop()

        backtrack(0, target)
        return result
````

## Permutations — Medium

**Pattern:** Backtracking + used set

**Recognition clues:** Need every ordering; order matters and each element is used once per path.

**Core idea:** At each recursion level, try every unused element, mark it used, recurse, then undo both path and used state.

**Complexity:** Time O(n·n!), recursion/used space O(n) excluding output.

**Common trap:** Undo **both** pieces of state: `path.pop()` and `used.remove(...)`.

````python
class Solution:
    def permute(self, nums):
        result = []
        path = []
        used = set()

        def backtrack():
            if len(path) == len(nums):
                result.append(path.copy())
                return

            for i, num in enumerate(nums):
                if i in used:
                    continue

                used.add(i)
                path.append(num)

                backtrack()

                path.pop()
                used.remove(i)

        backtrack()
        return result
````

# 15. Sorting / Custom Ordering

## Reorder Data in Log Files — Medium

**Pattern:** Custom sorting

**Recognition clues:** Letter-logs are sorted by content then identifier, while digit-logs keep original order and come after letter-logs.

**Core idea:** Separate letter/digit logs, sort only letter logs by `(content, identifier)`, then concatenate.

**Complexity:** Time O(n log n · L) approximately for sorting/comparisons, space O(n).

**Common trap:** Digit logs must remain stable in original order.

````python
class Solution:
    def reorderLogFiles(self, logs):
        letters = []
        digits = []

        for log in logs:
            identifier, rest = log.split(' ', 1)

            if rest[0].isdigit():
                digits.append(log)
            else:
                letters.append((rest, identifier, log))

        letters.sort(key=lambda x: (x[0], x[1]))

        return [log for _, _, log in letters] + digits
````

# 16. Prefix / Suffix Precomputation

## Product of Array Except Self — Medium

**Pattern:** Prefix / suffix products

**Recognition clues:** Each output depends on everything before and after an index, and division is disallowed.

**Core idea:** First store prefix product in answer, then multiply by a running suffix product from the right.

**Complexity:** Time O(n), auxiliary space O(1) excluding output.

**Common trap:** Update suffix after using it for the current position.

````python
class Solution:
    def productExceptSelf(self, nums):
        n = len(nums)
        answer = [1] * n

        prefix = 1
        for i in range(n):
            answer[i] = prefix
            prefix *= nums[i]

        suffix = 1
        for i in range(n - 1, -1, -1):
            answer[i] *= suffix
            suffix *= nums[i]

        return answer
````

# 17. Complexity Cheat Sheet

| Structure / operation | Typical complexity |
|---|---:|
| List index | O(1) |
| List append / pop end | Amortized O(1) |
| List membership | O(n) |
| Dict / set lookup | Average O(1) |
| Sort | O(n log n) |
| Heap push / pop | O(log n) |
| BFS / DFS graph | O(V + E) |
| Traverse grid | O(rows × cols) |
| Binary search | O(log n) |
| Nested full n × n loops | O(n²) |

### Complexity sanity test

If `n` is around:
- **10** → exponential may be okay.
- **20** → ~2ⁿ may be borderline but common for subsets.
- **10³** → O(n²) may sometimes pass.
- **10⁵** → usually target O(n) or O(n log n).
- **10⁶** → usually close to O(n).

Always use the actual problem constraints when available.

# 18. OA Debugging Checklist

### Arrays / strings
- Off-by-one in `range(...)`?

- Does sorting destroy needed original indices?

### Linked lists

- Would a dummy node simplify head deletion?

### Binary search
- Am I searching indices or an answer space?

### DP
- What exactly does `dp[i]` / `dp[r][c]` mean?
- Are base cases correct?
- Does every transition refer to already-computed states?

### Backtracking
- What changes before recursion?
- Did I undo every changed piece of state afterward?
- Do I need `start`, `used`, or duplicate skipping?

# 19. Pattern Comparison — Common Confusions

### BFS vs DFS
Use either for reachability/components. Prefer **BFS** when shortest unweighted distance or minute-by-minute layers matter.

### Sliding window vs two pointers
A sliding window usually represents a **contiguous range with state that changes as both boundaries move**. Two pointers more generally exploit order or pair relationships.

### Heap vs sorting
If you only need a small top-k / next minimum repeatedly, a heap can avoid sorting everything.

### DFS memoization vs bottom-up DP
They solve the same overlapping-subproblem idea in different directions:
- Top-down: write recursion naturally, cache states.
- Bottom-up: define computation order explicitly.

### Backtracking vs DP
Backtracking enumerates choices. DP is useful when many branches revisit the **same state** and only an aggregate answer is needed.

## Valid Sudoku — Medium

**Pattern:** Arrays & Hashing + Sets

**Recognition clues:** Need to validate repeated values across multiple constraints: each row, each column, and each 3×3 sub-box must contain no duplicates.

**Core idea:** Scan every cell and use sets to track values already seen in each row, column, and box. Identify a cell's box using `(r // 3, c // 3)`. If a value already exists in any corresponding set, return `False`.

**Complexity:** Time O(1), space O(1) because the board is always fixed at 9×9. Generalized to an n×n board, time and space would be O(n²).

**Common trap:** Remember to skip `"."` cells and check **all three constraints** — row, column, and 3×3 box. For the boxes, `(r // 3, c // 3)` gives a unique identifier for each one.

````python
from collections import defaultdict

class Solution:
    def isValidSudokuDefaultDict(self, board):
        rows = defaultdict(set)
        cols = defaultdict(set)
        boxes = defaultdict(set)

        for r in range(9):
            for c in range(9):
                value = board[r][c]

                if value == ".":
                    continue

                box = (r // 3, c // 3)

                if (
                    value in rows[r]
                    or value in cols[c]
                    or value in boxes[box]
                ):
                    return False

                rows[r].add(value)
                cols[c].add(value)
                boxes[box].add(value)

        return True

class Solution:
    def isValidSudoku(self, board: List[List[str]]) -> bool:
        boxes = {}

        for i in range(9):
            seenRow = set()
            seenColumn = set()

            for j in range(9):
                if board[i][j] in seenRow:
                    return False
                else:
                    if board[i][j] != '.':
                        seenRow.add(board[i][j])
                
                if board[j][i] in seenColumn:
                    return False
                else:
                    if board[j][i] != '.':
                        seenColumn.add(board[j][i])
                
                if board[i][j] != '.':
                    box = (i // 3, j // 3)

                    if box not in boxes:
                        boxes[box] = set()

                    if board[i][j] in boxes[box]:
                        return False

                    boxes[box].add(board[i][j])
        
        return True
````

## Longest Consecutive Sequence — Medium

**Pattern:** Hash Set + Sequence Start Detection

**Recognition clues:** Unsorted array, need to find consecutive values, and required O(n) time means sorting is too slow. Need O(1) average lookup to quickly check whether neighboring values exist.

**Core idea:** Convert `nums` to a set. A number is only the start of a sequence if `num - 1` does not exist in the set. Only from these starting numbers, repeatedly check `num + 1`, `num + 2`, etc. until the sequence ends, then update the longest length.

**Complexity:** Time O(n) average because each number is visited by the outer loop and the sequence-expansion work across all `while` loops is O(n) total. Hash set lookups are O(1) average. Space O(n) for the set.

**Common trap:** Do not expand a sequence from every number or assume the nested `for` + `while` automatically means O(n²). Only expand when `num - 1` does not exist; this ensures each consecutive sequence is walked once.

````python
class Solution:
    def longestConsecutive(self, nums: List[int]) -> int:
        numsSet = set(nums)
        longest = 0

        for num in numsSet:
            if num - 1 not in numsSet:
                length = 1

                while num + length in numsSet:
                    length += 1

                longest = max(longest, length)

        return longest
````

## Evaluate Reverse Polish Notation — Medium

**Pattern:** Stack

**Recognition clues:** Postfix expression where operators apply to the two most recently seen operands.

**Core idea:** Push numbers onto the stack. When an operator appears, pop b then a, compute a operator b, and push the result back.

**Complexity:** Time O(n), space O(n).

**Common trap:** Operand order matters for subtraction and division. In Python, use int(a / b) to truncate division toward zero.

````python
class Solution:
    def evalRPN(self, tokens: List[str]) -> int:
        stack = []
        operators = { '+', '*', '-', '/' }

        for token in tokens:
            if token not in operators:
                stack.append(int(token))
            else:
                y = stack.pop()
                x = stack.pop()

                if token == '+':
                    stack.append(x + y)
                elif token == '-':
                    stack.append(x - y)
                elif token == '*':
                    stack.append(x * y)
                else:
                    stack.append(int(x / y))
        
        return stack[0]
````

## Largest Rectangle in Histogram — Hard

**Pattern:** Monotonic increasing stack

**Recognition clues:** Need the largest rectangle across consecutive bars; a shorter bar determines when taller rectangles can no longer extend.

**Core idea:** Store (start_index, height) in an increasing stack. When a shorter bar appears, pop taller bars and calculate their areas. Carry the earliest popped index forward as the shorter bar's new start.

**Complexity:** Time O(n), space O(n).

**Common trap:** When popping a taller bar, update the current bar's start_index to the popped bar's index so the shorter bar can extend backward.

**Memory trigger:** Shorter bar arrives → taller rectangles are forced to end.

````python
class Solution:
    def largestRectangleArea(self, heights: List[int]) -> int:
        stack = []  # (start_index, height)
        max_area = 0

        for i, height in enumerate(heights):
            start = i

            while stack and stack[-1][1] > height:
                index, prev_height = stack.pop()
                max_area = max(max_area, prev_height * (i - index))
                start = index

            stack.append((start, height))

        for index, height in stack:
            max_area = max(max_area, height * (len(heights) - index))

        return max_area
        
        
````
